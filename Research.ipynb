```python
# ============================================================
# CRONBACH'S ALPHA ANALYSIS
# Dataset: ELogistics_Final_60_STATA_Format(1).xlsx
# ============================================================

import pandas as pd
import numpy as np
from google.colab import files

# ------------------------------------------------------------
# STEP 1: Upload the Excel file
# ------------------------------------------------------------
uploaded = files.upload("/content/ELogistics_Final_60_STATA_Format.xlsx")

file_path = "ELogistics_Final_60_STATA_Format(1).xlsx"

# ------------------------------------------------------------
# STEP 2: Read the STATA data sheet
# ------------------------------------------------------------
df = pd.read_excel(
    file_path,
    sheet_name="stata_data"
)

print("Dataset shape:", df.shape)
print("\nAvailable columns:")
print(df.columns.tolist())

# ------------------------------------------------------------
# STEP 3: Define Cronbach's alpha function
# ------------------------------------------------------------
def cronbach_alpha(data):
    """
    Calculate Cronbach's alpha.

    Parameters
    ----------
    data : pandas DataFrame
        DataFrame containing only the scale items.

    Returns
    -------
    float
        Cronbach's alpha value.
    """

    # Convert all selected columns to numeric
    data = data.apply(pd.to_numeric, errors="coerce")

    # Remove rows containing missing values
    data = data.dropna()

    # Number of scale items
    number_of_items = data.shape[1]

    if number_of_items < 2:
        raise ValueError(
            "Cronbach's alpha requires at least two items."
        )

    # Variance of individual items
    item_variances = data.var(axis=0, ddof=1)

    # Variance of the total scale score
    total_score = data.sum(axis=1)
    total_variance = total_score.var(ddof=1)

    if total_variance == 0:
        raise ValueError(
            "The total-score variance is zero. Alpha cannot be calculated."
        )

    alpha = (
        number_of_items / (number_of_items - 1)
    ) * (
        1 - item_variances.sum() / total_variance
    )

    return alpha

# ------------------------------------------------------------
# STEP 4: Select the e-logistics effectiveness scale items
# ------------------------------------------------------------
effectiveness_items = ["pi", "ci", "sr", "cv"]

effectiveness_data = df[effectiveness_items].copy()

# ------------------------------------------------------------
# STEP 5: Calculate Cronbach's alpha
# ------------------------------------------------------------
alpha_value = cronbach_alpha(effectiveness_data)

print("\n" + "=" * 55)
print("CRONBACH'S ALPHA RESULT")
print("=" * 55)
print("Construct: E-logistics Effectiveness")
print("Items:", effectiveness_items)
print("Number of items:", len(effectiveness_items))
print("Valid responses:", effectiveness_data.dropna().shape[0])
print(f"Cronbach's alpha: {alpha_value:.4f}")

# ------------------------------------------------------------
# STEP 6: Interpretation
# ------------------------------------------------------------
if alpha_value >= 0.90:
    interpretation = "Excellent internal consistency"
elif alpha_value >= 0.80:
    interpretation = "Good internal consistency"
elif alpha_value >= 0.70:
    interpretation = "Acceptable internal consistency"
elif alpha_value >= 0.60:
    interpretation = "Questionable internal consistency"
elif alpha_value >= 0.50:
    interpretation = "Poor internal consistency"
else:
    interpretation = "Unacceptable internal consistency"

print("Interpretation:", interpretation)
print("=" * 55)

# ------------------------------------------------------------
# STEP 7: Descriptive statistics of the items
# ------------------------------------------------------------
print("\nItem descriptive statistics:")
print(
    effectiveness_data.describe()
    .T[["count", "mean", "std", "min", "max"]]
    .round(3)
)
```
